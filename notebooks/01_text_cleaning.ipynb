## Imputing the `Text`

Due to some of the text are messages of the scraper being blocked by the website, we would standardize everything so that it contains the title, description, keywords, and the news snippet.

```python
type(news_data.iloc[0])
```

```python
def fill_text(row: pd.Series) -> str:
    # Get each field, convert to string, skip empty/NaN
    parts = [str(val).strip() for val in [row['title'], row['description'], row['keywords'], row['snippet']] if val and str(val).strip()]
    return ' '.join(parts)
```

```python
# tqdm message
tqdm.pandas(desc="Fixing text column")

# apply to the dataset
news_data["text"] = news_data.progress_apply(fill_text, axis=1)
```
